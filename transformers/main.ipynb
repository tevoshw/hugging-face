# Auto Tokenizer
Allows get a ready tokenizer trained from a model



## 1. Get the tokenizer
Get a tokenizer already trained and prepared

```python
from transformers import AutoTokenizer



tokenizer = AutoTokenizer.from_pretrained('model')

```
---

## 1.1 Types

| Model | AutoTokenizer | Vocab Size | Algorithm | Special Tokens |
|---|---|---:|---|---|
| **GPT-2** | `AutoTokenizer.from_pretrained("openai-community/gpt2")` | `50,257` | Byte-Level BPE | `<endoftext>` |
| **BERT** | `AutoTokenizer.from_pretrained("google-bert/bert-base-uncased")` | `30,522` | WordPiece | `[UNK]`, `[SEP]`, `[PAD]`, `[CLS]`, `[MASK]` |
| **RoBERTa** | `AutoTokenizer.from_pretrained("FacebookAI/roberta-base")` | `50,265` | Byte-Level BPE | `<s>`, `</s>`, `<unk>`, `<pad>`, `<mask>` |
| **T5** | `AutoTokenizer.from_pretrained("google-t5/t5-small")` | `32,100` | SentencePiece | `<unk>`, `<pad>`, `</s>`, `<extra_id_n>` |
| **LLaMA 3.2** | `AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-3B")` | `128,256` | BPE | `<begin_of_text> `, `<end_of_text>`, `<start_header_id>`, `<end_header_id>`, `<eot_id>` |
| **Qwen 2.5** | `AutoTokenizer.from_pretrained("Qwen/Qwen2.5-7B")` | `151,643` | BPE | `<endoftext>`, `<im_start>`, `<im_end>` |
| **GPT-Neo** | `AutoTokenizer.from_pretrained("EleutherAI/gpt-neo-125M")` | `50,257` | Byte-Level BPE | `<endoftext>` |

---

````python
from transformers import AutoTokenizer


tokenizer = AutoTokenizer.from_pretrained("gpt2")
````

---
## 2. Tokenize
Tokenize the text but maintain into string

```python

tokenizer.tokenize(text)

```

````python
output_tokenize = tokenizer.tokenize('Hello World')
print(output_tokenize)
````

---
## 3. Encode
Tokenize the text into IDS

```python

tokenizer.tokenize(text)

```

````python
output_encode = tokenizer.encode('Hello World')
print(output_encode)
````

---
## 4. Decode
Tokenize IDS into text


```python

tokenizer.tokenize(text)

```

````python
output_decode = tokenizer.decode(output_encode)
print(output_decode)
````


# 5. AutoTokenizer Settings

`AutoTokenizer` provides a high-level interface for loading and using
pre-trained tokenizers from the Hugging Face Hub.


---

## 1. Vocabulary

Returns the number of tokens in the tokenizer vocabulary.

### 1.1 Size of Vocab

```python
tokenizer.vocab_size
```

### 1.2 All the vocab

Returns the complete vocabulary as a dictionary:

```python
tokenizer.get_vocab()
```

````python
print(f'Vocab size: {tokenizer.vocab_size}')
print(f'All the vocab: {tokenizer.get_vocab()}')
````




---

# 2. Special Tokens

### 2.1 All Special tokens

Returns all special tokens defined by the tokenizer.

```python
tokenizer.all_special_tokens
```

### 2.2 All special IDS

Returns the IDs of all special tokens.

```python
tokenizer.all_special_ids
```


### 2.3  Special Tokens MAP

Returns the relationship between special-token roles and their tokens.

```python
tokenizer.special_tokens_map
```


### 2.4 Search Special token

Search the IDS of a special token, need the name


```python

tokenizer.name
```

````python
print(f'All special tokens: {tokenizer.all_special_tokens}')
print(f'All special IDS: {tokenizer.all_special_ids}')
print(f'Special tokens MAP: {tokenizer.special_tokens_map}')
print(f'EOS token: {tokenizer.eos_token_id}')
````
